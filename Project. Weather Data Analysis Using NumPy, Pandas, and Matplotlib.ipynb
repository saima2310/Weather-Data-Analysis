```python
import pandas as pd

# Part 2: Data Cleaning

# Load the dataset
df = pd.read_csv('climate_records.csv')

# 1. Check for missing (null) values in each column
print("--- Missing Values Before Cleaning ---")
print(df.isnull().sum())
```

```python
# 2. Handle missing values 
# Filling numerical columns with their respective mean values
df['Temperature_C'] = df['Temperature_C'].fillna(df['Temperature_C'].mean())
df['Humidity_Percent'] = df['Humidity_Percent'].fillna(df['Humidity_Percent'].mean())
df['Precipitation_mm'] = df['Precipitation_mm'].fillna(df['Precipitation_mm'].mean())

# Verify if any null values are left
print("\n--- Missing Values After Cleaning ---")
print(df.isnull().sum())

# 3. Check for duplicate rows and remove them
print(f"\nNumber of duplicate rows before removal: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"Number of duplicate rows after removal: {df.duplicated().sum()}")

# 4. Convert the 'Record_Date' column to proper datetime format
df['Record_Date'] = pd.to_datetime(df['Record_Date'])

# Verify data types after conversion
print("\n--- Data Types After Cleaning ---")
print(df.dtypes)
```

```python
import pandas as pd
import numpy as np

# Part 3: Descriptive Statistics with NumPy

# Load the cleaned dataset
df = pd.read_csv('climate_records.csv')

# Handle missing values first so calculations are accurate
df['Temperature_C'] = df['Temperature_C'].fillna(df['Temperature_C'].mean())
df['Humidity_Percent'] = df['Humidity_Percent'].fillna(df['Humidity_Percent'].mean())
df['Precipitation_mm'] = df['Precipitation_mm'].fillna(df['Precipitation_mm'].mean())

# 1. Temperature Statistics
temp_mean = np.mean(df['Temperature_C'])
temp_median = np.median(df['Temperature_C'])
temp_std = np.std(df['Temperature_C'])

print("--- Temperature Statistics (°C) ---")
print(f"Mean: {temp_mean:.2f}")
print(f"Median: {temp_median:.2f}")
print(f"Standard Deviation: {temp_std:.2f}")

# 2. Humidity Statistics
humidity_mean = np.mean(df['Humidity_Percent'])
humidity_median = np.median(df['Humidity_Percent'])
humidity_std = np.std(df['Humidity_Percent'])

print("\n--- Humidity Statistics (%) ---")
print(f"Mean: {humidity_mean:.2f}")
print(f"Median: {humidity_median:.2f}")
print(f"Standard Deviation: {humidity_std:.2f}")

# 3. Precipitation Statistics
precip_mean = np.mean(df['Precipitation_mm'])
precip_median = np.median(df['Precipitation_mm'])
precip_std = np.std(df['Precipitation_mm'])

print("\n--- Precipitation Statistics (mm) ---")
print(f"Mean: {precip_mean:.2f}")
print(f"Median: {precip_median:.2f}")
print(f"Standard Deviation: {precip_std:.2f}")
```

```python
import pandas as pd

# Part 4: Monthly and Seasonal Analysis

# Load the cleaned dataset
df = pd.read_csv('climate_records.csv')

# Ensure missing values are handled and date is in datetime format
df['Temperature_C'] = df['Temperature_C'].fillna(df['Temperature_C'].mean())
df['Humidity_Percent'] = df['Humidity_Percent'].fillna(df['Humidity_Percent'].mean())
df['Precipitation_mm'] = df['Precipitation_mm'].fillna(df['Precipitation_mm'].mean())
df['Record_Date'] = pd.to_datetime(df['Record_Date'])

# Extract month from the date column
df['Month'] = df['Record_Date'].dt.month

# Group by month and calculate the average (mean) values for each metric
monthly_avg = df.groupby('Month')[['Temperature_C', 'Humidity_Percent', 'Precipitation_mm']].mean()

print("--- Average Weather Metrics by Month ---")
print(monthly_avg)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Part 5: Visualizations with Matplotlib

# Load the cleaned dataset
df = pd.read_csv('climate_records.csv')

# Handle missing values and convert date
df['Temperature_C'] = df['Temperature_C'].fillna(df['Temperature_C'].mean())
df['Humidity_Percent'] = df['Humidity_Percent'].fillna(df['Humidity_Percent'].mean())
df['Precipitation_mm'] = df['Precipitation_mm'].fillna(df['Precipitation_mm'].mean())
df['Record_Date'] = pd.to_datetime(df['Record_Date'])

# 1. Line Plot: Temperature Trend Over Time
plt.figure(figsize=(10, 5))
plt.plot(df['Record_Date'], df['Temperature_C'], marker='o', color='b', linestyle='-')
plt.title('Temperature Trend Over Time')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 2. Bar Plot: Average Precipitation by Month
df['Month'] = df['Record_Date'].dt.month
monthly_precip = df.groupby('Month')['Precipitation_mm'].mean()

plt.figure(figsize=(8, 5))
monthly_precip.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Precipitation by Month')
plt.xlabel('Month')
plt.ylabel('Precipitation (mm)')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 3. Histogram: Temperature Distribution
plt.figure(figsize=(8, 5))
plt.hist(df['Temperature_C'], bins=10, color='orange', edgecolor='black', alpha=0.7)
plt.title('Temperature Distribution')
plt.xlabel('Temperature (°C)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# Part 6: Identify Extreme Weather

# Load the cleaned dataset
df = pd.read_csv('climate_records.csv')

# Handle missing values and convert date
df['Temperature_C'] = df['Temperature_C'].fillna(df['Temperature_C'].mean())
df['Humidity_Percent'] = df['Humidity_Percent'].fillna(df['Humidity_Percent'].mean())
df['Precipitation_mm'] = df['Precipitation_mm'].fillna(df['Precipitation_mm'].mean())
df['Record_Date'] = pd.to_datetime(df['Record_Date'])

# Convert columns to NumPy arrays for logical filtering
temperatures = df['Temperature_C'].to_numpy()
precipitation = df['Precipitation_mm'].to_numpy()
dates = df['Record_Date'].to_numpy()

# 1. Define thresholds for extreme weather
# Extreme Heat: Temperature > 35°C
# Heavy Rain: Precipitation > 10.0 mm
extreme_temp_threshold = 35.0
extreme_rain_threshold = 10.0

# 2. Use NumPy boolean indexing to filter extreme days
extreme_heat_indices = np.where(temperatures > extreme_temp_threshold)
heavy_rain_indices = np.where(precipitation > extreme_rain_threshold)

print("--- Extreme Temperature Days (> 35°C) ---")
extreme_heat_df = df.iloc[extreme_heat_indices]
print(extreme_heat_df[['Record_Date', 'Temperature_C', 'Humidity_Percent']])

print("\n--- Heavy Precipitation Days (> 10 mm) ---")
heavy_rain_df = df.iloc[heavy_rain_indices]
print(heavy_rain_df[['Record_Date', 'Precipitation_mm']])
```
